```python
# Define Course Knowledge Base
courses = {
    "AI": {
        "Beginner": ["AI Basics", "Intro to ML"],
        "Advanced": ["Deep Learning", "Neural Networks"]
    },
    "Data Science": {
        "Beginner": ["Data Science Basics", "Python for Analysis"],
        "Advanced": ["Machine Learning", "Big Data Analytics"]
    }
}

def recommend_courses(interest, level):
    # Normalize input to match dictionary keys
    interest = interest.title()
    level = level.capitalize()

    if interest == "Ai":
        interest = "AI"

    # Rule-Based Decision Logic
    if interest in courses and level in courses[interest]:
        return courses[interest][level]
    else:
        return ["No match found. Try again."]

# Main Execution Flow
user_interest = input("Enter Interest (AI / Data Science): ")
user_level = input("Enter Level (Beginner / Advanced): ")

print(recommend_courses(user_interest, user_level))
```
